## 01. XGBoost Baseline for Protein Stability Prediction

This notebook builds the initial tabular baseline for predicting mutation-induced protein stability changes, measured as ΔΔG in kcal/mol. The goal is to test how far mutation-level physicochemical features can go before adding structural graph information.

It produces the XGBoost baseline outputs used in the final blog and saves reusable CSV files for downstream plotting.

**single_point_mutations.tsv**: contains only single point mutations (2531 rows)

**1temp.tsv**: contains all ProThermDB entries w/ $\Delta\Delta G$ values (3949 rows, not included in repo but potential future addition)

## 6. XGBoost baseline formulation

We train XGBoost as a baseline.

Input: mutation-level feature vector  
Output: experimentally measured ΔΔG  
Objective: squared-error regression  
Metrics: RMSE and Pearson correlation

XGBoost is a useful baseline because it can learn nonlinear feature interactions while still using only flat mutation-level information. Keep only df columns that will be used by XGBoost

```python
cols_to_keep = ['PROTEIN', '∆∆G_(kcal/mol)', 'delta_hydrophobicity', 'delta_charge', 'delta_weight']

one_hot_cols = [c for c in table_uniprot_xgb.columns 
                if (c.startswith('WT_') or c.startswith('MUT_')) 
                and c not in ['WT_AA', 'MUT_AA']]  # ← exclude raw string columns

cols_to_keep += one_hot_cols
df = table_uniprot_xgb[cols_to_keep].copy()
baseline_uniprot_df = df.rename(columns={'∆∆G_(kcal/mol)': 'ddG'})

print(f"dataset shape: {baseline_uniprot_df.shape}")
print(f"unique proteins: {baseline_uniprot_df['PROTEIN'].nunique()}")
```

## 7. Train/Validation/Test splits

We compare two splitting strategies.

The per-mutation split distributes mutations from the same protein across train, validation, and test sets. This tests interpolation within proteins the model has already partially seen.

The protein-level split holds out entire proteins for validation and testing. This is the stricter and more realistic test of generalization to unseen proteins.

### 1. Per-mutation split (for each protein, distribute its mutations 60/20/20 across train/val/test)

```python
def per_protein_split(baseline_uniprot_df, train_frac=0.6, val_frac=0.2, seed=42):
    rng = np.random.default_rng(seed)
    train_idx, val_idx, test_idx = [], [], []

    for protein, group in baseline_uniprot_df.groupby('PROTEIN'):
        idx = group.index.tolist()
        rng.shuffle(idx)
        n = len(idx)
        n_train = max(1, int(np.floor(train_frac * n)))
        n_val   = max(1, int(np.floor(val_frac * n)))
        # remaining go to test (at least 1 if possible)
        train_idx += idx[:n_train]
        val_idx   += idx[n_train:n_train + n_val]
        test_idx  += idx[n_train + n_val:]

    return train_idx, val_idx, test_idx

train_idx, val_idx, test_idx = per_protein_split(df)

train_df = df.loc[train_idx]
val_df   = df.loc[val_idx]
test_df  = df.loc[test_idx]

print(f"\nPer-mutation split:")
print(f"  Train: {len(train_df)} ({len(train_df)/len(df):.1%})")
print(f"  Val:   {len(val_df)}   ({len(val_df)/len(df):.1%})")
print(f"  Test:  {len(test_df)}  ({len(test_df)/len(df):.1%})")
```

```python
train_df.columns
```

### 2. Protein level split (entire proteins assigned to one split only; to test unseeen proteins)

```python
def protein_level_split(baseline_uniprot_df, train_frac=0.6, val_frac=0.2, seed=42):
    rng = np.random.default_rng(seed)
    proteins = baseline_uniprot_df['PROTEIN'].unique().tolist()
    rng.shuffle(proteins)

    n = len(proteins)
    n_train = int(np.floor(train_frac * n))
    n_val = int(np.floor(val_frac * n))

    train_proteins = set(proteins[:n_train])
    val_proteins = set(proteins[n_train:n_train + n_val])
    test_proteins = set(proteins[n_train + n_val:])

    return (
        baseline_uniprot_df[baseline_uniprot_df['PROTEIN'].isin(train_proteins)],
        baseline_uniprot_df[baseline_uniprot_df['PROTEIN'].isin(val_proteins)],
        baseline_uniprot_df[baseline_uniprot_df['PROTEIN'].isin(test_proteins)],
    )

train_prot, val_prot, test_prot = protein_level_split(df)

print(f"\nProtein-level split:")
print(f"  Train: {len(train_prot)} mutations, {train_prot['PROTEIN'].nunique()} proteins")
print(f"  Val:   {len(val_prot)} mutations,   {val_prot['PROTEIN'].nunique()} proteins")
print(f"  Test:  {len(test_prot)} mutations,  {test_prot['PROTEIN'].nunique()} proteins")
```

### 3. Data splits

```python
feature_cols = [c for c in baseline_uniprot_df.columns if c not in ['PROTEIN', 'ddG']]

X_train = train_df[feature_cols].values
y_train = train_df['∆∆G_(kcal/mol)'].values

X_val   = val_df[feature_cols].values
y_val   = val_df['∆∆G_(kcal/mol)'].values

X_test  = test_df[feature_cols].values
y_test  = test_df['∆∆G_(kcal/mol)'].values

print(f"Features used: {feature_cols}")
print(f"X_train shape: {X_train.shape}")
```

### 4. Train XGBoost

```python
from xgboost import XGBRegressor
from sklearn.metrics import mean_squared_error
from scipy.stats import pearsonr
import numpy as np
```

```python
# sanity check model; final saved XGBoost outputs use validation selected 'best_cfg'
model = XGBRegressor(
    n_estimators=300, learning_rate=0.05, max_depth=4, subsample=0.8, colsample_bytree=0.8,
    random_state=42, eval_metric='rmse', early_stopping_rounds=20,)

model.fit(X_train, y_train, eval_set=[(X_val, y_val)], verbose=50)
```

## 8. Hyperparameter search

We tune XGBoost hyperparameters using validation-set performance.

The purpose of this section is not to exhaustively optimize XGBoost, but to select a reasonable baseline configuration before comparing split behavior.

```python
from itertools import product

configs = [
    {'max_depth': 3, 'learning_rate': 0.05, 'n_estimators': 300},
    {'max_depth': 4, 'learning_rate': 0.05, 'n_estimators': 300},
    {'max_depth': 6, 'learning_rate': 0.05, 'n_estimators': 300},
    {'max_depth': 4, 'learning_rate': 0.01, 'n_estimators': 300},
    {'max_depth': 4, 'learning_rate': 0.10, 'n_estimators': 300},
]

print("── Hyperparameter Search (Val only) ─────────────────")
print(f"{'depth':>6} {'lr':>6} | {'Val r':>8} {'Val RMSE':>10}")
print("-" * 40)

hyper_rows = []

for cfg in configs:
    m = XGBRegressor(
        **cfg,
        subsample=0.8,
        colsample_bytree=0.8,
        random_state=42,
        eval_metric="rmse",
        early_stopping_rounds=20,
    )

    m.fit(X_train, y_train, eval_set=[(X_val, y_val)], verbose=False)

    val_preds = m.predict(X_val)
    val_r, _ = pearsonr(y_val, val_preds)
    val_rmse = np.sqrt(mean_squared_error(y_val, val_preds))

    hyper_rows.append({
        "max_depth": cfg["max_depth"],
        "learning_rate": cfg["learning_rate"],
        "n_estimators": cfg["n_estimators"],
        "val_r": val_r,
        "val_rmse": val_rmse,
        "best_iteration": m.best_iteration,
    })

    print(f"  {cfg['max_depth']:>4} {cfg['learning_rate']:>6} | {val_r:>8.3f} {val_rmse:>10.3f}")

xgb_hyperparameter_search = pd.DataFrame(hyper_rows)
best_row = xgb_hyperparameter_search.sort_values("val_r", ascending=False).iloc[0]

best_cfg = {
    "max_depth": int(best_row["max_depth"]),
    "learning_rate": float(best_row["learning_rate"]),
    "n_estimators": int(best_row["n_estimators"]),
}

if SAVE_RESULTS:
    xgb_hyperparameter_search.to_csv(XGB_RESULTS_DIR / f"xgb_hyperparameter_search_{XGB_RUN_TAG}.csv", index=False)

display(xgb_hyperparameter_search)
print(f"\n→ Best config: {best_cfg}")
```

```python
import json

best_config_payload = {
    "best_cfg": best_cfg,
    "selection_metric": "validation Pearson r",
    "selected_from": "per-mutation validation split",
}

with open(XGB_RESULTS_DIR / f"xgb_best_config_{XGB_RUN_TAG}.json", "w") as f:
    json.dump(best_config_payload, f, indent=2)

print("Saved best config:", best_config_payload)
```

## 9. XGBoost evaluation

We evaluate the selected XGBoost model on both per-mutation and protein-level splits.

The key comparison is whether performance remains stable when the model must generalize to unseen proteins.

```python
def evaluate(model, X, y, split_name):
    preds = model.predict(X)
    rmse = np.sqrt(mean_squared_error(y, preds))
    r, _ = pearsonr(y, preds)
    print(f"{split_name:10s} | Pearson r: {r:.3f} | RMSE: {rmse:.3f} kcal/mol")
    return preds, r, rmse

print("── Results ──────────────────────────────")
_, _, _  = evaluate(model, X_train, y_train, "Train")
_, _, _  = evaluate(model, X_val,   y_val,   "Val")
test_preds, test_r, test_rmse = evaluate(model, X_test, y_test, "Test")
```

```python
fig, axes = plt.subplots(1, 4, figsize=(22, 5))

split_configs = {
    'Per-Mutation': (X_train, y_train, X_val, y_val, X_test, y_test, 20),
    'Protein-Level': (
        train_prot[feature_cols].values, train_prot['∆∆G_(kcal/mol)'].values,
        val_prot[feature_cols].values,   val_prot['∆∆G_(kcal/mol)'].values,
        test_prot[feature_cols].values,  test_prot['∆∆G_(kcal/mol)'].values,
        50  # more patience for protein-level since val set is small
    )
}

colors = {'Per-Mutation': 'steelblue', 'Protein-Level': 'coral'}
curves = {}
final_preds = {}

for split_name, (*data, patience) in split_configs.items():
    Xtr, ytr, Xv, yv, Xt, yt = data
    m = XGBRegressor(
        **best_cfg, subsample=0.8, colsample_bytree=0.8,
        random_state=42, eval_metric='rmse',
        early_stopping_rounds=patience
    )
    m.fit(Xtr, ytr, eval_set=[(Xtr, ytr), (Xv, yv)], verbose=False)
    results = m.evals_result()
    curves[split_name] = {
        'train': results['validation_0']['rmse'],
        'val':   results['validation_1']['rmse'],
        'best':  m.best_iteration
    }
    preds = m.predict(Xt)
    final_preds[split_name] = {
        'test_preds': preds,
        'y_test': yt,
        'r':    pearsonr(yt, preds)[0],
        'rmse': np.sqrt(mean_squared_error(yt, preds))
    }

# ── Panel A: Per-Mutation training curve ──────────────────────
for i, (split_name, col) in enumerate(colors.items()):
    ax = axes[i]
    c = curves[split_name]
    ax.plot(c['train'], color=col, linestyle='-',
            label='Train RMSE', linewidth=5)
    ax.plot(c['val'],   color=col, linestyle='--',
            label='Val RMSE',   linewidth=5, alpha=0.8)
    ax.axvline(c['best'], color='gray', linestyle=':',
               linewidth=3, label=f'Best round ({c["best"]})')
    ax.set_xlabel('Boosting Round', fontsize=20)
    ax.set_ylabel('RMSE (kcal/mol)', fontsize=20)
    ax.set_title(f'{split_name} Split\nTraining Curve', fontsize=20)
    ax.legend(fontsize=12, frameon=False)
    ax.tick_params(axis='both', labelsize=20)
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

# ── Panels C & D: Predicted vs actual ─────────────────────────
for i, (split_name, col) in enumerate(colors.items()):
    ax = axes[i + 2]
    d = final_preds[split_name]
    ax.scatter(d['y_test'], d['test_preds'],
               alpha=0.5, s=100, color=col, edgecolors='none')
    lims = [min(d['y_test'].min(), d['test_preds'].min()) - 0.5,
            max(d['y_test'].max(), d['test_preds'].max()) + 0.5]
    ax.plot(lims, lims, 'k--', linewidth=1, label='Perfect prediction')
    ax.set_xlabel('True ΔΔG (kcal/mol)', fontsize=20)
    ax.set_ylabel('Predicted ΔΔG (kcal/mol)', fontsize=20)
    ax.set_title(f'{split_name} Split: Test Set\n'
                 f'Pearson r = {d["r"]:.3f}\n'
                 f'RMSE = {d["rmse"]:.3f} kcal/mol', fontsize=20)
    ax.legend(fontsize=12, frameon=False)
    ax.spines['top'].set_visible(False)
    ax.tick_params(axis='both', labelsize=20)
    ax.spines['right'].set_visible(False)

# plt.suptitle('XGBoost Baseline: Training Dynamics and Test Performance',
#              fontsize=13, fontweight='bold', y=1.02)
plt.tight_layout()

if SAVE_FIGURES:
    fig.savefig(ASSETS_DIR / f"xgb_training_and_test_performance_{XGB_RUN_TAG}.png", dpi=300, bbox_inches="tight")

plt.show()
```

## 11. Save baseline outputs

```python
xgb_metric_rows = []
xgb_prediction_rows = []
xgb_curve_rows = []
xgb_importance_rows = []

split_configs = {
    "Per-Mutation": {
        "train": (X_train, y_train, train_df),
        "val": (X_val, y_val, val_df),
        "test": (X_test, y_test, test_df),
        "patience": 20,
    },
    "Protein-Level": {
        "train": (train_prot[feature_cols].values, train_prot["∆∆G_(kcal/mol)"].values, train_prot),
        "val": (val_prot[feature_cols].values, val_prot["∆∆G_(kcal/mol)"].values, val_prot),
        "test": (test_prot[feature_cols].values, test_prot["∆∆G_(kcal/mol)"].values, test_prot),
        "patience": 50,
    },
}

for split_name, split_data in split_configs.items():
    Xtr, ytr, train_frame = split_data["train"]
    Xv, yv, val_frame = split_data["val"]
    Xt, yt, test_frame = split_data["test"]

    m = XGBRegressor(
        **best_cfg,
        subsample=0.8,
        colsample_bytree=0.8,
        random_state=42,
        eval_metric="rmse",
        early_stopping_rounds=split_data["patience"],
    )

    m.fit(Xtr, ytr, eval_set=[(Xtr, ytr), (Xv, yv)], verbose=False)

    evals = m.evals_result()
    for epoch, rmse in enumerate(evals["validation_0"]["rmse"]):
        xgb_curve_rows.append({
            "split_strategy": split_name,
            "set": "train",
            "epoch": epoch,
            "rmse": rmse,
            "best_iteration": m.best_iteration,
        })

    for epoch, rmse in enumerate(evals["validation_1"]["rmse"]):
        xgb_curve_rows.append({
            "split_strategy": split_name,
            "set": "val",
            "epoch": epoch,
            "rmse": rmse,
            "best_iteration": m.best_iteration,
        })

    for set_name, (X, y, frame) in {
        "train": (Xtr, ytr, train_frame),
        "val": (Xv, yv, val_frame),
        "test": (Xt, yt, test_frame),
    }.items():
        preds = m.predict(X)
        rmse = np.sqrt(mean_squared_error(y, preds))
        r = pearsonr(y, preds)[0] if len(y) > 2 else np.nan

        xgb_metric_rows.append({
            "model": "XGBoost",
            "split_strategy": split_name,
            "set": set_name,
            "rmse": rmse,
            "r": r,
            "n": len(y),
            "best_iteration": m.best_iteration,
            "max_depth": best_cfg["max_depth"],
            "learning_rate": best_cfg["learning_rate"],
            "n_estimators": best_cfg["n_estimators"],
        })

        pred_frame = frame.copy().reset_index(drop=False).rename(columns={"index": "original_index"})
        pred_frame["model"] = "XGBoost"
        pred_frame["split_strategy"] = split_name
        pred_frame["set"] = set_name
        pred_frame["y_true"] = y
        pred_frame["y_pred"] = preds
        pred_frame["residual"] = pred_frame["y_pred"] - pred_frame["y_true"]
        xgb_prediction_rows.append(pred_frame)

    for feat, importance in zip(feature_cols, m.feature_importances_):
        xgb_importance_rows.append({
            "split_strategy": split_name,
            "feature": feat,
            "importance": importance,
        })

xgb_metrics = pd.DataFrame(xgb_metric_rows)
xgb_predictions = pd.concat(xgb_prediction_rows, ignore_index=True)
xgb_training_curves = pd.DataFrame(xgb_curve_rows)
xgb_feature_importance = pd.DataFrame(xgb_importance_rows)

if SAVE_RESULTS:
    xgb_metrics.to_csv(XGB_RESULTS_DIR / f"xgb_split_metrics_{XGB_RUN_TAG}.csv", index=False)
    xgb_predictions.to_csv(XGB_RESULTS_DIR / f"xgb_predictions_{XGB_RUN_TAG}.csv", index=False)
    xgb_training_curves.to_csv(XGB_RESULTS_DIR / f"xgb_training_curves_{XGB_RUN_TAG}.csv", index=False)
    xgb_feature_importance.to_csv(XGB_RESULTS_DIR / f"xgb_feature_importance_{XGB_RUN_TAG}.csv", index=False)
    table_uniprot_xgb.to_csv(XGB_RESULTS_DIR / f"xgb_processed_uniprot_data_{XGB_RUN_TAG}.csv", index=False)

display(xgb_metrics)
print("Saved XGBoost outputs to:", XGB_RESULTS_DIR)
```
